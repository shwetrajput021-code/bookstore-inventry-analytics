```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
inventory = pd.DataFrame({
    "Title": ["Python Basics", "Data Analysis",
              "The Silent River", "Business Fundamentals",
              "Money Matters"],
    "Author": ["A Shah", "R Patel", "M Joshi", "K Mehta",
               "S Desai"],
    "Genre": ["Technology", "Technology", "Fiction", "Business",
              "Finance"],
    "Price": [450, 650, 300, 500, 400],
    "Quantity": [20, 15, 25, 18, 22]
})

print(inventory)
inventory.to_csv("inventory.csv", index=False)

print("Inventory CSV file saved successfully!")
```

```python
sales = pd.DataFrame({
    "Date":[
        "2026-10-01","2026-10-02","2026-10-03","2026-10-04","2026-10-05"
        ],
    "Title":[
        "Python Basics","Data Analysis","The Silent River","Business Fundamentals","Money Matters"
        ],
    "Quantity Sold": [2, 3, 1, 2, 4],
    "Total Revenue": [900, 1950, 300, 1000, 1600]
})

print(sales)
sales.to_csv("sales.csv", index=False)
print("sales.csv saved successfully!")
```

```python
inventory_data = pd.read_csv("inventory.csv")
sales_data = pd.read_csv("sales.csv")

print("BOOK INVENTORY")
print(inventory_data)

print("\nSALES DATA")
print(sales_data)
```

```python
total_revenue = sales_data["Total Revenue"].sum()
total_sold = sales_data["Quantity Sold"].sum()

print("Total Revenue:", total_revenue)
print("Total Books Sold:", total_sold)
```

```python
best_selling = sales_data.groupby("Title")["Quantity Sold"].sum()

print(best_selling)
```

```python
best_book = sales_data.groupby("Title")["Quantity Sold"].sum().idxmax()

print("best selling book : ", best_book)
```

```python
total_stock = inventory_data["Quantity"].sum()
print("Total Stock:", total_stock)
```

```python
class Bookstore:
    def __init__(self):
        self.books = inventory_data

    def show_books(self):
        print(self.books)

store = Bookstore()
store.show_books()
```

```python
def add_book(title, author, genre, price, quantity):
    new_book = {"Title": title,"Author": author,"Genre": genre,
                "Price": price,"Quantity": quantity
                
                }

    global inventory_data
    inventory_data.loc[len(inventory_data)] = new_book

    print("Book Added Successfully!")
    
add_book("Python Advanced", "Rahul", "Education", 500, 10)
```

```python
def record_sale(title, quantity, price):
    total = quantity * price

    print("Book:", title)
    print("Quantity Sold:", quantity)
    print("Total Revenue:", total)
    
record_sale("Python Basics", 2, 450)
```

```python
def generate_report():
    total_revenue = sales_data["Total Revenue"].sum()
    total_sold = sales_data["Quantity Sold"].sum()

    print("Total Revenue: ₹", total_revenue)
    print("Total Books Sold:", total_sold)
    
generate_report()
```

```python
plt.bar(sales_data["Title"], sales_data["Quantity Sold"])
plt.title("Book Sales Report")
plt.xlabel("Book Name")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=sales_data, x="Title", y="Total Revenue")
plt.title("Revenue by Book")
plt.xticks(rotation=45)
plt.show()
```
